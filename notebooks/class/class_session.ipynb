```python
import pandas as pd
import numpy as np

 
# Q1. Create a DataFrame


students = {
    "Roll_No": [101,102,103,104,105,106,107,108,109,110],
    "Name": ["Aman","Riya","Karan","Neha","Rahul","Sneha","Arjun","Pooja","Vikas","Anjali"],
    "Branch": ["CSE","IT","ECE","CSE","ME","IT","ECE","CSE","ME","IT"],
    "Marks": [85,78,92,67,74,88,59,95,81,76]
}

df = pd.DataFrame(students)

print(df.head())
print(df.tail())


 
# Q2. Series Operations
 

marks = pd.Series([85,78,92,67,74,88,59,95,81,76])

print("Maximum:", marks.max())
print("Minimum:", marks.min())
print("Mean:", marks.mean())
print("Median:", marks.median())


 
# Q3. DataFrame Information
 

employee = {
    "Emp_ID":[1,2,3,4,5],
    "Name":["Amit","Priya","Rohan","Neha","Sahil"],
    "Department":["HR","IT","Sales","Finance","IT"],
    "Salary":[45000,60000,52000,58000,65000]
}

emp_df = pd.DataFrame(employee)

print("Shape:")
print(emp_df.shape)

print("\nColumns:")
print(emp_df.columns)

print("\nData Types:")
print(emp_df.dtypes)

print("\nStatistical Summary:")
print(emp_df.describe())


 
# Q4. Selecting Columns
 

print(df[["Name","Branch","Marks"]])


 
# Q5. Selecting Rows
 

print("First 10 Rows")
print(df.iloc[:10])

print("\nRows from index 5 to 15")
print(df.iloc[5:16])


 
# Q6. Using loc
 

df_loc = df.set_index("Roll_No")

print(df_loc)

print(df_loc.loc[[101,105,109]])


 
# Q7. Filtering Data
 

result = df[df["Marks"] > 75]

print(result)


 
# Q8. Multiple Conditions
 

result = df[(df["Branch"]=="CSE") & (df["Marks"]>80)]

print(result)


 
# Q9. Adding a Column
 

df["Result"] = np.where(df["Marks"]>=40,"Pass","Fail")

print(df)

# Q10. Sorting Data


sorted_df = df.sort_values(by="Marks",ascending=False)

print(sorted_df)

print("\nTop 5 Students")

print(sorted_df.head())
```
