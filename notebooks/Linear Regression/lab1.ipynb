```python
# Linear Regression 
# Linear regression is a statistical method used to model the relationship between a dependent variable and one or more independent variables. It assumes that the relationship between the variables can be approximated by a straight line. The goal of linear regression is to find the best-fitting line that minimizes the sum of the squared differences between the observed values and the predicted values.


# Multiple Linear Regression
# Multiple linear regression is an extension of simple linear regression that allows for the inclusion of multiple independent variables. It is used when the relationship between the dependent variable and multiple independent variables needs to be modeled. The goal is to find the best-fitting hyperplane that minimizes the sum of the squared differences between the observed values and the predicted values.

#In simple linear regression we have 2D graphs, while in multiple linear regression we have 3D graphs. In simple linear regression, we have one independent variable and one dependent variable, while in multiple linear regression, we have multiple independent variables and one dependent variable.


# MAE (Mean Absolute Error)
# Mean Absolute Error (MAE) is a metric used to evaluate the accuracy of a regression model. It measures the average absolute difference between the predicted values and the actual values. A lower MAE indicates a better fit of the model to the data.

#RMSE (Root Mean Squared Error)
# Root Mean Squared Error (RMSE) is another metric used to evaluate the accuracy of a regression model. It measures the square root of the average squared differences between the predicted values and the actual values. RMSE gives more weight to larger errors, making it sensitive to outliers. A lower RMSE indicates a better fit of the model to the data.

#Gradient Descent
# Gradient Descent is an optimization algorithm used to minimize the cost function in machine learning models, including linear regression. It iteratively adjusts the model parameters (weights) in the direction of the steepest descent of the cost function, which is typically the mean squared error in regression problems. The learning rate determines the size of the steps taken towards the minimum, and careful tuning of this parameter is essential for convergence.

#Global minima 
# In the context of optimization, a global minimum refers to the point in the parameter space where the cost function reaches its lowest possible value. In linear regression, this corresponds to the set of model parameters that result in the best fit to the data, minimizing the error between predicted and actual values. Achieving a global minimum ensures that the model is as accurate as possible given the data.



```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
df = pd.read_csv('../../data/price.csv')
```

```python
df
```

```python
X =df[['area']]
y =df[['price']]
```

```python
model = LinearRegression()
model.fit(X, y)
```

```python
y_pred = model.predict(X)
y_pred
```

```python

plt.scatter(X, y_pred, color='blue')
plt.plot(X, y_pred, color='red')
```

```python
model.coef_
```

```python
model.intercept_
```

```python
model.predict([[1000]])
```

```python
print("Mean Squared Error:", mean_squared_error(y, y_pred))
```
