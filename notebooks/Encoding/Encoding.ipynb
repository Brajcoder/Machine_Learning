# Encoding

```python
# Ordinal Encoding
from sklearn.preprocessing import OrdinalEncoder
ord_enc = OrdinalEncoder(categories=[
    ['Poor', 'Average', 'Good', 'Excellent'],
    ['School', 'UG', 'PG', 'PhD']
])


```

```python
X_train = [["Quality", "Education"],
           ["Good", "UG"],
          ]
```

```python
ord_enc.fit(X_train)  # Fit the encoder to the training data
```

```python
Y_train = ["Good", "UG", "Excellent", "PG", "Average", "School"]
```

```python
# Label Encoding
from sklearn.preprocessing import LabelEncoder
label_enc = LabelEncoder()
label_enc.fit(Y_train)  # Fit the encoder to the training data
print(label_enc.transform(Y_train))  # Transform the training data
```
