```python
import pandas as pd
import numpy as np

students_missing = {
    "Name": ["Aman", "Deepak", "Neha", "Karan", "Aditya"],
    "Marks": [85, np.nan, 78, 65, np.nan],
    "Age": [20, 21, np.nan, 22, 20]
}

df_missing = pd.DataFrame(students_missing)

print(df_missing)

print("Using isnull():")
print(df_missing.isnull())

```

```python
df_missing["Marks"] = df_missing["Marks"].fillna(
    df_missing["Marks"].mean()
)

df_missing["Age"] = df_missing["Age"].fillna(
    df_missing["Age"].mean()
)

print(df_missing)
```

```python
students_duplicate = {
    "Name": ["Aman", "Deepak", "Neha", "Aman", "Karan", "Deepak"],
    "Roll_No": [101, 102, 103, 101, 104, 102],
    "Marks": [85, 92, 78, 85, 65, 92]
}

df_duplicate = pd.DataFrame(students_duplicate)

print("Before removing duplicates:")
print(df_duplicate)

df_duplicate = df_duplicate.drop_duplicates()

print("After removing duplicates:")
print(df_duplicate)
```
