```python

arr38 = np.array([10, 3, 7, 12, 17, 20, 23, 25, 29])

prime_numbers = []

for num in arr38:
    if num > 1:
        for i in range(2, int(np.sqrt(num)) + 1):
            if num % i == 0:
                break
        else:
            prime_numbers.append(num)
print(prime_numbers)
```
