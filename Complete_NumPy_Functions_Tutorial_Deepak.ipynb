## 1. Array creation

The most common constructors are `np.array`, `np.zeros`, `np.ones`, `np.full`, `np.arange`, and `np.linspace`. Use `arange` when you know the step size; use `linspace` when you know the number of evenly spaced values.


```python
# From Python sequences
a = np.array([1, 2, 3, 4])
b = np.array([[1, 2, 3], [4, 5, 6]], dtype=np.float64)

# Special-purpose constructors
zeros = np.zeros((2, 3), dtype=int)
ones = np.ones((2, 3))
filled = np.full((2, 2), 7)
identity = np.eye(3)
range_values = np.arange(0, 10, 2)
line_values = np.linspace(0, 1, 5)

for name, value in {"a": a, "b": b, "zeros": zeros, "ones": ones, "filled": filled, "identity": identity, "range_values": range_values, "line_values": line_values}.items():
    print(f"{name}:\n{value}\n")
```

## 2. Inspecting arrays

These attributes and functions answer the basic questions about an array: what it contains, how it is shaped, and how much memory it uses.

| Property or function | Meaning |
|---|---|
| `ndim` | Number of axes |
| `shape` | Length along each axis |
| `size` | Total number of elements |
| `dtype` | Element data type |
| `itemsize` | Bytes used by one element |
| `nbytes` | Total raw bytes used by elements |
| `astype` | Converts to another data type |


```python
x = np.array([[10, 20, 30], [40, 50, 60]], dtype=np.int32)
print("ndim:", x.ndim)
print("shape:", x.shape)
print("size:", x.size)
print("dtype:", x.dtype)
print("itemsize:", x.itemsize)
print("nbytes:", x.nbytes)
print("as float:\n", x.astype(np.float64))
```

## 3. Indexing, slicing, and filtering

NumPy uses zero-based indexing. A slice usually creates a **view** rather than a copy, so changing a slice can change the original array. Use `.copy()` when independent data is required. Boolean masks are the standard way to filter values.


```python
x = np.arange(1, 13).reshape(3, 4)
print("x =\n", x)
print("row 0:", x[0])
print("element at row 1, column 2:", x[1, 2])
print("first two rows, last two columns:\n", x[:2, -2:])
print("reversed columns:\n", x[:, ::-1])

mask = x % 2 == 0
print("even values:", x[mask])
print("values greater than 7:", x[x > 7])

labels = np.where(x >= 8, "high", "low")
print("labels:\n", labels)
```

## 4. Reshaping and combining arrays

`reshape` changes the view when possible without changing the data. `ravel` returns a flattened view when possible, while `flatten` always returns a copy. Stacking functions combine arrays along existing or new axes.


```python
a = np.arange(1, 7)
print("reshape:\n", a.reshape(2, 3))
print("transpose:\n", a.reshape(2, 3).T)
print("ravel:", a.reshape(2, 3).ravel())
print("flatten:", a.reshape(2, 3).flatten())

left = np.array([[1, 2], [3, 4]])
right = np.array([[5, 6], [7, 8]])
print("concatenate rows:\n", np.concatenate([left, right], axis=0))
print("concatenate columns:\n", np.concatenate([left, right], axis=1))
print("vstack:\n", np.vstack([left, right]))
print("hstack:\n", np.hstack([left, right]))
print("split:", np.array_split(np.arange(10), 3))
```

## 5. Sorting, searching, and set operations

NumPy includes efficient routines for ordering data and locating values. `argsort` returns positions rather than sorted values, which is useful for sorting related arrays by the same key.


```python
values = np.array([7, 2, 9, 2, 5])
print("sorted:", np.sort(values))
print("sort indices:", np.argsort(values))
print("unique values and counts:", np.unique(values, return_counts=True))
print("where value equals 2:", np.where(values == 2)[0])
print("searchsorted in sorted array:", np.searchsorted(np.sort(values), 6))
print("intersection:", np.intersect1d([1, 2, 3], [2, 3, 4]))
print("union:", np.union1d([1, 2, 3], [3, 4, 5]))
```

## 6. Arithmetic and universal functions

NumPy applies operations element by element. Universal functions, or **ufuncs**, provide optimized implementations of common mathematical operations and support broadcasting.


```python
x = np.array([1, 4, 9, 16], dtype=float)
print("x + 2:", x + 2)
print("x * 3:", x * 3)
print("x squared:", np.square(x))
print("square root:", np.sqrt(x))
print("natural log:", np.log(x))
print("exponential:", np.exp(np.array([0, 1, 2])))
print("sin(pi/2):", np.sin(np.pi / 2))
print("rounded values:", np.round(np.array([1.234, 5.678]), 2))
print("absolute values:", np.abs(np.array([-3, 0, 4])))
```

### Broadcasting

Broadcasting allows arrays with compatible shapes to participate in one operation without manually copying values. Dimensions are compatible when they are equal or when one of them is `1`.


```python
prices = np.array([[10, 20, 30], [12, 22, 32]])
tax_rates = np.array([0.05, 0.10, 0.15])
final_prices = prices * (1 + tax_rates)
print("prices:\n", prices)
print("tax rates:", tax_rates)
print("final prices:\n", final_prices)

try:
    np.ones((2, 3)) + np.ones((2, 2))
except ValueError as error:
    print("Incompatible shapes raise ValueError:", error)
```

## 7. Aggregations and descriptive statistics

Aggregations reduce many values to a summary. Supply `axis=0` to reduce down rows, `axis=1` to reduce across columns, or omit `axis` to reduce the entire array.


```python
data = np.array([[12, 15, 18], [20, 22, 25], [8, 10, 14]])
print("sum:", np.sum(data))
print("mean:", np.mean(data))
print("median:", np.median(data))
print("minimum / maximum:", np.min(data), np.max(data))
print("standard deviation:", np.std(data))
print("variance:", np.var(data))
print("column means:", np.mean(data, axis=0))
print("row sums:", np.sum(data, axis=1))
print("argmin / argmax:", np.argmin(data), np.argmax(data))
print("cumulative sum:", np.cumsum(np.array([1, 2, 3, 4])))
print("percentiles:", np.percentile(data, [25, 50, 75]))
```

## 8. Comparisons, logic, and conditional selection

Functions such as `all`, `any`, `logical_and`, `logical_or`, and `logical_not` make conditions explicit. `np.select` is useful when several conditions map to different results.


```python
scores = np.array([45, 67, 82, 91, 38])
passed = scores >= 50
excellent = scores >= 85
print("passed:", passed)
print("everyone passed?", np.all(passed))
print("at least one excellent score?", np.any(excellent))

conditions = [scores >= 85, scores >= 50]
choices = ["excellent", "pass"]
print(np.select(conditions, choices, default="fail"))
print("clipped:", np.clip(scores, 50, 90))
```

## 9. Missing values and data types

Floating-point arrays can represent missing values with `np.nan`. Use NaN-aware functions such as `nanmean` and `nansum` when missing values should be ignored. For exact labels or mixed values, consider masked arrays or a higher-level data library.


```python
measurements = np.array([10.0, 12.5, np.nan, 15.0, np.nan])
print("is NaN:", np.isnan(measurements))
print("regular mean:", np.mean(measurements))
print("NaN-aware mean:", np.nanmean(measurements))
print("NaN-aware sum:", np.nansum(measurements))

integer_values = np.array([1, 2, 3], dtype=np.int64)
boolean_values = np.array([True, False, True])
print("integer dtype:", integer_values.dtype)
print("boolean sum counts True values:", boolean_values.sum())
```

## 10. Linear algebra

The `numpy.linalg` namespace contains common matrix and vector operations. `@` performs matrix multiplication, while `*` performs element-wise multiplication.


```python
A = np.array([[3.0, 1.0], [1.0, 2.0]])
b = np.array([9.0, 8.0])

print("matrix multiplication:\n", A @ A)
print("determinant:", np.linalg.det(A))
print("inverse:\n", np.linalg.inv(A))
print("eigenvalues:", np.linalg.eigvals(A))
print("vector norm:", np.linalg.norm(b))
print("solution to Ax=b:", np.linalg.solve(A, b))
print("least-squares solution:", np.linalg.lstsq(A, b, rcond=None)[0])
```

## 11. Random numbers and reproducibility

Use the modern `default_rng` interface for new code. A local generator keeps random state explicit and reproducible without changing global random state.


```python
rng = np.random.default_rng(Deepak_seed := 42)
print("uniform samples:", rng.uniform(0, 1, size=5))
print("normal samples:", rng.normal(loc=0, scale=1, size=5))
print("random integers:", rng.integers(1, 10, size=5))
print("choice without replacement:", rng.choice(np.arange(10), size=4, replace=False))

# The same seed produces the same sequence.
rng1 = np.random.default_rng(42)
rng2 = np.random.default_rng(42)
print("reproducible:", np.array_equal(rng1.random(4), rng2.random(4)))
```

## 12. File input and output

For simple numerical arrays, `save` and `load` preserve NumPy's binary format. `savetxt` and `loadtxt` are useful when a human-readable text file is required.


```python
from pathlib import Path

output_path = Path("numpy_demo_array.npy")
original = np.arange(6).reshape(2, 3)
np.save(output_path, original)
restored = np.load(output_path)
print("restored array:\n", restored)
print("same values:", np.array_equal(original, restored))
output_path.unlink(missing_ok=True)
```

## 13. Performance: vectorization versus Python loops

Vectorized operations express an operation on the whole array and usually avoid Python-level iteration. The exact speed improvement depends on hardware, array size, and operation; benchmark your own workload with `timeit`.


```python
import timeit

values = np.arange(1_000_000, dtype=np.float64)
loop_time = timeit.timeit("[value * 2 for value in values]", globals={"values": values}, number=3)
vector_time = timeit.timeit("values * 2", globals={"values": values}, number=3)
print(f"Python loop total: {loop_time:.4f} seconds")
print(f"NumPy vectorized total: {vector_time:.4f} seconds")
print(f"Approximate speed ratio: {loop_time / vector_time:.1f}x")
```

## 14. Common mistakes and debugging patterns

Shape errors are among the most frequent NumPy problems. Inspect `shape` and `dtype` before applying an operation. Remember that a one-dimensional array has shape `(n,)`, not `(n, 1)`; use `[:, None]` or `reshape(-1, 1)` when a column vector is required.


```python
row = np.array([1, 2, 3])
column = row[:, None]
print("row shape:", row.shape)
print("column shape:", column.shape)
print("outer-style multiplication:\n", column * row)

# A safe assertion documents the shape expected by downstream code.
assert column.shape == (3, 1)
print("Shape assertion passed.")
```
